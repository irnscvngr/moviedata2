```python
import pandas as pd
```

### Letterboxd Watchlist

```python
df_watchlist = pd.read_csv(
    "data/watchlist-ironscavenger-2026-08-21-08-35-utc.csv"
    )

df_watchlist.columns = [col.lower() for col in df_watchlist.columns]

df_watchlist
```

### All movies and runtimes

```python
# ----------------------------------------
# Get all movies
df_all_movies = pd.read_csv(
    "data/omdb/all_movies.csv",
    usecols=range(4),
    )

# ----------------------------------------
# Get all movie details
df_movie_details = pd.read_csv(
    "data/omdb/movie_details.csv",
    usecols=range(4),
)

# ----------------------------------------
# Get all movie votes
df_movie_votes = pd.read_csv(
    "data/omdb/all_votes.csv",
    usecols=range(3),
    )

# ----------------------------------------
# Join details with movies
df_all_movies = (
    df_all_movies
    .merge(
        df_movie_details,
        how='left',
        left_on='id',
        right_on='movie_id'
        )
    .drop(columns=['id','parent_id'])
)

# ----------------------------------------
# Join votes with movies
df_all_movies = (
    df_all_movies
    .merge(
        df_movie_votes,
        how='left',
        on='movie_id'
    )
)

# ----------------------------------------
# Convert date-column to actual datetime

# Replace "\N" entries with actual Nulls
sel = df_all_movies['date'] == "\\N"
df_all_movies.loc[sel,'date'] = pd.NA
# Convert values
df_all_movies['date'] = pd.to_datetime(df_all_movies['date'], format='%Y-%m-%d')

df_all_movies
```

### Movie categories

```python
# ----------------------------------------
# Get all movie categories
df_movie_categories = pd.read_csv(
    "data/omdb/movie_categories.csv",
    usecols=range(2),
    )

# ----------------------------------------
# Get category names
df_category_names = pd.read_csv(
    "data/omdb/category_names.csv",
    usecols=range(3),
    )

# ----------------------------------------
# Only select english category names and drop language column
df_category_names = (
    df_category_names[df_category_names['language_iso_639_1']=='en']
    .drop(columns='language_iso_639_1')
)

# ----------------------------------------
# Join category names with IDs
df_movie_categories = (
    df_movie_categories
    .merge(
        df_category_names,
        on='category_id',
        how='left'
    )
    .rename(columns={'name':'category_name'})
)

# ----------------------------------------
# Join movie names with categories
df_movie_categories = (
    df_movie_categories
    .merge(
        df_all_movies[['movie_id', 'name']],
        on='movie_id'
    )
    .sort_values(['movie_id','category_id'])
)

df_movie_categories
```
